```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample dataset
data = {
    'Customer_ID': [1, 2, 3, 4, 5, 6, 7],
    'Age': [22, 34, 26, 19, 45, 23, 31],
    'Product': ['Electronics', 'Clothing', 'Electronics', 'Electronics', 'Groceries', 'Clothing', 'Electronics'],
    'Purchase_Value': [9500, 2500, 8800, 7600, 1200, 3000, 9100]
}

# Create DataFrame
df = pd.DataFrame(data)

# Descriptive Analysis: Electronics purchases by age group
electronics = df[df['Product'] == 'Electronics']

# 1. Count and percentage by age group
age_18_25 = electronics[(electronics['Age'] >= 18) & (electronics['Age'] <= 25)]
percentage = (len(age_18_25) / len(electronics)) * 100

print("Total Electronics Purchases:", len(electronics))
print("Purchases by Age 18-25:", len(age_18_25))
print(f"Percentage (18-25 age group): {percentage:.2f}%")

# 2. Average Purchase Value for this group
avg_value = age_18_25['Purchase_Value'].mean()
print(f"Average Purchase Value (18-25 age group): ₹{avg_value:.2f}")

# Optional: Plot product distribution
product_counts = df['Product'].value_counts()
product_counts.plot(kind='bar', title='Product Purchase Distribution')
plt.xlabel('Product')
plt.ylabel('Number of Purchases')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Simulated dataset
data = {
    'Customer_ID': range(1, 11),
    'Age': [22, 34, 26, 19, 45, 23, 31, 29, 38, 52],
    'Gender': ['F', 'M', 'F', 'F', 'M', 'F', 'M', 'M', 'F', 'M'],
    'Product': ['Phone', 'Laptop', 'Shoes', 'Phone', 'Groceries', 'Shoes', 'Laptop', 'Phone', 'Groceries', 'Phone'],
    'Quantity': [1, 1, 2, 1, 5, 1, 1, 2, 3, 1],
    'Price': [10000, 45000, 1500, 10000, 200, 1500, 45000, 10000, 200, 10000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate total value
df['Total_Value'] = df['Quantity'] * df['Price']

# --- 1. Total sales per product ---
product_sales = df.groupby('Product')['Total_Value'].sum().sort_values(ascending=False)
print("Total Sales by Product:")
print(product_sales)

# --- 2. Most purchased product ---
most_purchased = df['Product'].value_counts().idxmax()
print("\nMost Purchased Product:", most_purchased)

# --- 3. Average sales per product ---
avg_sales = df.groupby('Product')['Total_Value'].mean()
print("\nAverage Sales per Product:")
print(avg_sales)

# --- 4. Sales by Age Group ---
bins = [0, 18, 25, 35, 50, 100]
labels = ['<18', '18-25', '26-35', '36-50', '50+']
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)

age_group_sales = df.groupby('Age_Group')['Total_Value'].sum()
print("\nSales by Age Group:")
print(age_group_sales)

# --- 5. Gender-based total sales ---
gender_sales = df.groupby('Gender')['Total_Value'].sum()
print("\nSales by Gender:")
print(gender_sales)

# --- 6. Plots ---
# Product sales
product_sales.plot(kind='bar', title='Total Sales by Product', color='darkgreen')
plt.ylabel('₹ Total Sales')
plt.tight_layout()
plt.show()

# Sales by Age Group
sns.barplot(x=age_group_sales.index, y=age_group_sales.values, palette='magma')
plt.title('Sales by Age Group')
plt.ylabel('₹ Total Sales')
plt.xlabel('Age Group')
plt.tight_layout()
plt.show()

# Gender comparison
gender_sales.plot(kind='pie', autopct='%1.1f%%', title='Sales Distribution by Gender', legend=False)
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create synthetic dataset
np.random.seed(42)
n = 1000

df = pd.DataFrame({
    'Amount': np.random.exponential(scale=100, size=n).round(2),
    'Time': np.random.randint(0, 24, size=n),
    'Transaction_Type': np.random.choice(['Online', 'POS', 'ATM'], size=n, p=[0.5, 0.4, 0.1]),
    'Location': np.random.choice(['Urban', 'Suburban', 'Rural'], size=n, p=[0.6, 0.3, 0.1]),
    'Is_Fraud': np.random.choice([0, 1], size=n, p=[0.95, 0.05])
})

# 2. Preprocessing
X = df.drop('Is_Fraud', axis=1)
y = df['Is_Fraud']

# One-hot encoding categorical features
X_encoded = pd.get_dummies(X, columns=['Transaction_Type', 'Location'], drop_first=True)

# Feature scaling
scaler = StandardScaler()
X_encoded[['Amount', 'Time']] = scaler.fit_transform(X_encoded[['Amount', 'Time']])

# 3. Train/test split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.3, stratify=y, random_state=42)

# 4. Model training
model = LogisticRegression()
model.fit(X_train, y_train)

# 5. Prediction and evaluation
y_pred = model.predict(X_test)

print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, digits=4))

# 6. Optional: Visualizing Confusion Matrix
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# 1. Generate synthetic dataset
np.random.seed(42)
n = 1000

df = pd.DataFrame({
    'Amount': np.random.exponential(scale=100, size=n).round(2),
    'Time': np.random.randint(0, 24, size=n),
    'Transaction_Type': np.random.choice(['Online', 'POS', 'ATM'], size=n, p=[0.5, 0.4, 0.1]),
    'Location': np.random.choice(['Urban', 'Suburban', 'Rural'], size=n, p=[0.6, 0.3, 0.1]),
    'Is_Fraud': np.random.choice([0, 1], size=n, p=[0.95, 0.05])  # 5% fraud
})

# 2. Descriptive Analysis: Pie Chart
fraud_counts = df['Is_Fraud'].value_counts()
labels = ['Legit (0)', 'Fraud (1)']
colors = ['skyblue', 'salmon']
explode = [0, 0.1]

plt.figure(figsize=(6, 6))
plt.pie(fraud_counts, labels=labels, autopct='%1.1f%%', startangle=140,
        colors=colors, explode=explode, shadow=True)
plt.title('💳 Fraud vs Legit Transactions')
plt.tight_layout()
plt.show()

# 3. Preprocessing
X = df.drop('Is_Fraud', axis=1)
y = df['Is_Fraud']

# One-hot encode categorical features
X = pd.get_dummies(X, columns=['Transaction_Type', 'Location'], drop_first=True)

# Scale numerical features
scaler = StandardScaler()
X[['Amount', 'Time']] = scaler.fit_transform(X[['Amount', 'Time']])

# Train/Test split
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.3, random_state=42)

# 4. Build Neural Network
model = Sequential([
    Dense(32, activation='relu', input_shape=(X_train.shape[1],)),
    Dropout(0.3),
    Dense(16, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 5. Train model
history = model.fit(X_train, y_train, epochs=20, batch_size=32, validation_split=0.2, verbose=0)

# 6. Evaluate
y_pred = (model.predict(X_test) > 0.5).astype(int).flatten()

print("\n📊 Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\n📄 Classification Report:\n", classification_report(y_test, y_pred, digits=4))

# 7. Plot training history
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('📈 Model Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.tight_layout()
plt.show()
```
