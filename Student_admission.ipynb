```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# 1. Generate synthetic dataset
np.random.seed(42)
n = 1000

GPA = np.round(np.random.uniform(6.0, 10.0, n), 2)
GRE = np.random.randint(280, 340, n)
TOEFL = np.random.randint(90, 120, n)
SOP = np.round(np.random.uniform(1, 5, n), 1)
LOR = np.round(np.random.uniform(1, 5, n), 1)
research = np.random.randint(0, 2, n)
internship = np.random.randint(0, 2, n)
tier = np.random.randint(1, 4, n)

# Admission based on rules
admit = ((GPA > 8.0) & (GRE > 310) & (TOEFL > 100) & (SOP > 3.5) & (LOR > 3.5)).astype(int)
admit = np.where((GRE > 330) | ((GPA > 9.0) & research == 1), 1, admit)
admit = np.where((GPA < 7.0) | (TOEFL < 95), 0, admit)

df = pd.DataFrame({
    'GPA': GPA,
    'GRE': GRE,
    'TOEFL': TOEFL,
    'SOP': SOP,
    'LOR': LOR,
    'Research': research,
    'Internship': internship,
    'University_Tier': tier,
    'Admitted': admit
})

# 2. Train model
X = df.drop('Admitted', axis=1)
y = df['Admitted']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

model = LogisticRegression()
model.fit(X_train, y_train)

# 3. Take user input
print("\n📋 Enter Student Details:")
gpa = float(input("GPA (0–10): "))
gre = int(input("GRE Score (out of 340): "))
toefl = int(input("TOEFL Score (out of 120): "))
sop = float(input("SOP Strength (1–5): "))
lor = float(input("LOR Strength (1–5): "))
research = int(input("Research Experience (1=Yes, 0=No): "))
internship = int(input("Internship Done (1=Yes, 0=No): "))
tier = int(input("University Tier (1=Top, 2=Mid, 3=Low): "))

student_input = np.array([[gpa, gre, toefl, sop, lor, research, internship, tier]])
student_scaled = scaler.transform(student_input)
prediction = model.predict(student_scaled)[0]

# 4. Explanation logic
reasons = []

if gpa < 7.0:
    reasons.append("Low GPA (<7.0)")
elif gpa >= 9.0:
    reasons.append("Excellent GPA")

if gre < 310:
    reasons.append("GRE below average (<310)")
elif gre > 330:
    reasons.append("Excellent GRE score")

if toefl < 95:
    reasons.append("Low TOEFL (<95)")
elif toefl > 105:
    reasons.append("Strong TOEFL score")

if sop < 3.0:
    reasons.append("Weak SOP (<3.0)")
elif sop > 4.0:
    reasons.append("Strong SOP")

if lor < 3.0:
    reasons.append("Weak LOR (<3.0)")
elif lor > 4.0:
    reasons.append("Strong LOR")

if research:
    reasons.append("Has research experience")
else:
    reasons.append("No research experience")

if internship:
    reasons.append("Completed internship")
else:
    reasons.append("No internship experience")

if tier == 1:
    reasons.append("Targeting top-tier university")
elif tier == 3:
    reasons.append("Targeting lower-tier university")

# 5. Show Result
print("\n🎯 Prediction:", "✅ Admitted" if prediction == 1 else "❌ Not Admitted")
print("\n📌 Reasoning based on input:")
for r in reasons:
    print(" -", r)
```
